```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import re

# Load your data — if from file
# df = pd.read_csv("your_data.tsv", sep="\t")

# Or paste directly as a string (this example assumes you've saved it in 'data.tsv')
df = pd.read_csv("mfs.kstar.results.tsv", sep="\t")

# Extract mutation label from the Sequence column
def extract_mutation_label(seq):
    # extract just the mutations in order of B3, B5, B9, B14
    match = re.findall(r'B3=\w+|B5=\w+|B9=\w+|B14=\w+', seq)
    sorted_match = sorted(match, key=lambda x: int(x[1:x.find('=')]))  # sort by number (B3, B5, etc.)
    return ', '.join(sorted_match)

df['Mutation'] = df['Sequence'].apply(extract_mutation_label)

# Sort by K* upper bound for readability
df = df.sort_values(by='K* Upper (log10)', ascending=True)

# Plotting
x = np.arange(len(df))
width = 0.35

fig, ax = plt.subplots(figsize=(14, 7))
bar1 = ax.bar(x - width/2, df['K* Lower (log10)'], width, label='Lower Bound', color='steelblue')
bar2 = ax.bar(x + width/2, df['K* Upper (log10)'], width, label='Upper Bound', color='orange')

ax.set_ylabel('log₁₀ K* Score')
ax.set_title('K* Score (log₁₀) for Mutations at B3, B5, B9, B14')
ax.set_xticks(x)
ax.set_xticklabels(df['Mutation'], rotation=90)
ax.legend()
plt.tight_layout()
plt.show()
```
